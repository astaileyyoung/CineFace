Dispersion represents the distance between a given face and the median center of faces. It's calculated as so:

```python
centers = df.groupby('imdb_id')[['center_x_norm', 'center_y_norm']].median()
df = df.join(centers, on='imdb_id', rsuffix='_med')
df['disp'] = np.sqrt((df['center_x_norm'] - df['center_x_norm_med'])**2 + 
                        (df['center_y_norm'] - df['center_y_norm_med'])**2)
```
