### Avg. Face Size vs. Std. of Face Size Std.

This plot maps a director's average shot scale (X) against their cross-project variance (Y). While the x-axis shows the career-wide average face size, the y-axis measures how much those averages shift from one movie to the next. A high y-score indicates a director whose 'typical' shot size is highly *inconsistent* across their filmography. A low y-score indicates a director who is highly *consistent* across their filmography.

```python
x = "z_size_mv_g_mean"
y = "z_size_mv_g_std"
plot_directors_against_sample(g, 
                              x, 
                              y, 
                              [name], 
                              x_title="Avg. Face Size (Z Score)", 
                              y_title="Std. of std. Face Size", 
                              title="Christopher Nolan: Avg. Face Size vs. Std. of Avg. Face Size", 
                              width=1200,
                              height=600,
                              dst="./plots/nolan/nolan_size_mean_v_size_std_std.png")
```

```python
print("Top 1 Avg. Size:", stats.percentileofscore(g['z_size_mv_g_mean'], g[g['name'] == name].iloc[0]['z_size_mv_g_mean']))
print("Top 1 Std. of Avg. Size:", stats.percentileofscore(g['z_size_mv_g_std'], g[g['name'] == name].iloc[0]['z_size_mv_g_std']))
```

Nolan is relatively high on the std. of std. What does this mean? It means Nolan has changed his face scale considerably between films. 

### Avg. Size Variance vs. Std. of Size Variance 

This plot looks at how face sizes vary within a movie vs. how the variance varies between movies. The x axis measures the *internal* variance while the y axis measures the *external* variance. A high value on the x axis means the director has a lot of variation in shot scales *within* a film. A high value on the y axis means the director is "consistenly inconsistent." A representative example is Sergio Leone who likes to vary his shot scales within his films (think of the sparse landscapes contrasted with extreme closeups in a film like *The Good, the Bad, and the Ugly* [1966]). But he also varies shot scales *between* movies, meaning on some films he has a lot of variation in scales while on some films he maintains a more consistent scale. 

```python
x = "z_v_size_mv_g_mean"
y = "z_v_size_mv_g_std"
plot_directors_against_sample(g, 
                              x, 
                              y, 
                              [name], 
                              x_title="Std. Face Size (Z Score)", 
                              y_title="Std. of std. Face Size Variance (Z Score)", 
                              title="Christopher Nolan: Std. Face Size vs. Std. of Face Size Variance", 
                              width=1200,
                              height=600,
                              dst="./plots/nolan/nolan_size_mean_v_size_std.png")
```

```python
print("Std. Size:", stats.percentileofscore(g['z_v_size_mv_g_mean'], g[g['name'] == name].iloc[0]['z_v_size_mv_g_mean']))
print("Std. of Std. Size:", stats.percentileofscore(g['z_v_size_mv_g_std'], g[g['name'] == name].iloc[0]['z_v_size_mv_g_std']))
```

Nolan sits right on the regression line, suggesting a more "normal" pattern even though he is relatively high on both metrics. Nolan likes to vary his scales *between* films but in a way that is expected give his variation *within* his films.

## Cameras

```python
query = """
    SELECT *
    FROM vwCamerasByPerson
    WHERE person_name = 'Christopher Nolan'
        AND job_name = 'Director'
"""
camera_df = pd.read_sql(query, engine)
camera_df
```

```python
y = ["z_size_mv_g", "z_f_per_fr_mv_g", "z_gini_mv_g", "z_dist_mv_g"]

names = {
    "z_size_mv_g": "Avg. Size",
    "z_f_per_fr_mv_g": "Avg. Faces per Frame",
    "z_gini_mv_g": "Avg. Gini",
    "z_dist_mv_g": "Avg. Distance",
}

camera_g = camera_df[y + ["is_imax"]].groupby('is_imax').mean()
camera_g = camera_g.rename(names, axis=1)
camera_g.round(2)
```

```python
styled_table = (
    camera_g
    .style
    .format(precision=2)  # This is the key line for the image export
)

await dfi.export_async(styled_table, "./data/nolan/nolan_camera_matrix.png")
```

```python
query = """
    SELECT 
        is_imax,
        ROUND(AVG(z_size_mv_g), 2) AS `Avg. Size`,
        ROUND(AVG(z_f_per_fr_mv_g), 2) AS `Avg. Faces per Frame`,
        ROUND(AVG(z_gini_mv_g), 2) AS `Avg. Gini`,
        ROUND(AVG(z_dist_mv_g), 2) AS `Avg. Distance`
    FROM factWork fw 
    INNER JOIN bridgeCamera bc ON fw.work_id = bc.work_id
    INNER JOIN dimCamera dc ON bc.camera_id = dc.camera_id
    GROUP BY is_imax
"""
camera_sample = pd.read_sql(query, engine)
camera_sample
```

```python
styled_table = (
    camera_sample
    .style
    .format(precision=2)  # This is the key line for the image export
)

await dfi.export_async(styled_table, "./data/nolan/nolan_sample_camera_matrix.png")
```

## Personnel

```python
query = """
    WITH nolan AS (
        SELECT work_id 
        FROM `vwWorksByPerson`
        WHERE person_name = "Christopher Nolan"
            AND job_name = "Director"
    )
    SELECT 
        person_name,
        ROUND(AVG(z_size_mv_g), 2) AS `Avg. Face Size`,
        ROUND(AVG(z_f_per_fr_mv_g), 2) AS `Avg. Faces per Frame`,
        ROUND(AVG(z_vert_mv_g), 2) AS `Vertical Discipline`,
        ROUND(AVG(z_gini_mv_g), 2) AS `Gini`,
        ROUND(AVG(z_dist_mv_g), 2) AS `Avg. Distance`
    FROM `vwWorksByPerson` wbp
    INNER JOIN nolan n ON n.work_id = wbp.work_id
    WHERE person_name IN ("Wally Pfister", "Hoyte van Hoytema") AND wbp.work_id IN (n.work_id)
    GROUP BY person_name
"""
cinematographer_df = pd.read_sql(query, engine)
cinematographer_df
```
